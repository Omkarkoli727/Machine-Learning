Task 3: Outlier Detection

```python
plt.figure(figsize=(18,10))
sns.boxplot(data)
```

```python
data.columns
```

```python
plt.figure(figsize=(10,8))
sns.boxplot(data['Population'])
```

So we have Outlier in our Population column

```python
#for removing outliers
def roy(data, x):
    q1 = np.quantile(data[x], 0.25)
    q3 = np.quantile(data[x], 0.75)
    iqr = q3 - q1
    
    lowest_value = q1 - 1.5 * iqr
    highest_value = q3 + 1.5 * iqr
    
    filtered_data = data[(data[x] > lowest_value) & (data[x] < highest_value)]
    
    print("Removing outliers completed...")
    sns.boxplot(data[x])
    return filtered_data
```

```python
data=roy(data,"GDP")
 #run until the dots disappear
```

```python
data=roy(data,"Population") 
```

```python
data=roy(data,"infant deaths") 
```

Task 5 — Analytical Summary
What the dataset is telling you
Life expectancy (2000–2015) differs strongly by development Status, and the visuals show both a level gap and a trend (improvement over time). The correlation view supports what you’d expect in public health data: higher schooling and better income composition align with higher life expectancy, while adult mortality and HIV/AIDS align with lower life expectancy.

What to watch out for before modeling or deeper inference
Missingness is material in Population, Hepatitis B, and GDP. If you plan to do regression/ML, you’ll want a clear strategy (drop vs impute; potentially status- or region-aware imputation).

Outliers are not “bad data” here—they’re structural. Variables like GDP, infant deaths, and HIV/AIDS naturally have extreme country-years. For most analyses, you’ll get cleaner results by using IQR to remove the Outliers.
